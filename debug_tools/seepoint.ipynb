```python
from plyfile import PlyData
import numpy as np

path = "/home/jian/gaussian-splatting/output/mip360/bicycle/point_cloud/fcgb_pro/iteration_500/point_cloud.ply"

ply = PlyData.read(path)

print("=== PLY ELEMENTS ===")
for elem in ply.elements:
    print(f"\nElement name: {elem.name}")
    print(f"Count: {elem.count}")

    print("Properties:")
    for p in elem.properties:
        print(f"  - {p.name}: {p.dtype}")

    # 打印前 5 条数据
    data = elem.data
    print("First 5 entries:")
    for i in range(min(5, elem.count)):
        print(data[i])
```
