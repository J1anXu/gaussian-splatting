```python
import os
import numpy as np
from plyfile import PlyData, PlyElement

def seperate(iter):
    path = f"/home/jian/gaussian-splatting/output/mip360/bicycle/point_cloud/fcgb_pro_max/iteration_{iter}/point_cloud.ply"
    print(f"Processing iteration {iter}")
    ply = PlyData.read(path)

    vertex = ply['vertex'].data
    block_ids = vertex['block_id']

    out_dir = f"debug/blocks"
    os.makedirs(out_dir, exist_ok=True)

    unique_blocks = np.unique(block_ids)

    print(f"Found {len(unique_blocks)} blocks")

    for bid in [3,4]:
        mask = (block_ids == bid)
        sub_vertices = vertex[mask]

        print(f"Block {bid}: {len(sub_vertices)} points")

        el = PlyElement.describe(sub_vertices, 'vertex')
        out_ply = PlyData([el], text=ply.text)

        out_path = os.path.join(out_dir, f"block_{int(bid)}_{iter}.ply")
        out_ply.write(out_path)
```

```python

itearsions = [1700,1800,1900,2000,2100,2200,2300,2400,2500,2600,2700,2800,2900,3000,3100,3200,3300,3400,3500,3600,3700,3800,3900,4000]
for iter in itearsions:

    seperate(iter)
```
